```python
# -------------------------------------------------------------
# Step 0: Import Libraries and Set Random Seed
# -------------------------------------------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import datetime
import csv
import warnings

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, TimeDistributed
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error

# Optionally, suppress feature name warnings:
# warnings.filterwarnings("ignore", category=UserWarning)

# Set seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# -------------------------------------------------------------
# Step 1: Data Loading and Preprocessing
# -------------------------------------------------------------
# Load your Excel data file (adjust file path as needed)
data = pd.read_excel('/content/EPF_data_Indiv.xlsx')

# Expected columns:
#   Date | Price | Demand Day-ahead DE | Wind and PV Day ahead (MWh/h) | Gas | Coal | CO2
print("Raw data:")
print(data.head())

# Remove duplicate Date rows and convert "Date" to datetime
data.drop(data.loc[data['Date'].duplicated()].index, inplace=True)
data['Date'] = pd.to_datetime(data['Date'])

# Set "Date" as the index and drop the original column
data.set_index('Date', inplace=True)

# Filter to include only records from 2019-01-01 onward
data = data[data.index >= '2019-01-01']
print("Data after filtering (from 2019-01-01):")
print(data.head())

# (Optional) Add a time-of-day feature if needed:
# data['Hour'] = data.index.hour

# -------------------------------------------------------------
# Step 2: Define Target, Features, and Split by Year
# -------------------------------------------------------------
# According to your file:
#   - Target: "Price"
#   - Features: ["Demand Day-ahead DE", "Wind and PV Day ahead (MWh/h)", "Gas", "Coal", "CO2"]
target_column = "Price"
features_list = ["Demand Day-ahead DE", "Wind and PV Day ahead (MWh/h)", "Gas", "Coal", "CO2"]
print("Target column:", target_column)
print("Feature columns:", features_list)

# Get list of unique years
unique_years = sorted(data.index.year.unique())
print("Unique years in data:", unique_years)

# Use first four years for training (e.g., 2019–2022) and the remaining for testing
train_years = [year for year in unique_years if year <= 2022]
test_years  = [year for year in unique_years if year > 2022]
print("Training years:", train_years)
print("Test years:", test_years)

# Create training and test DataFrames
train_data = data[data.index.year.isin(train_years)]
test_data = data[data.index.year.isin(test_years)]
print("Training data shape:", train_data.shape)
print("Test data shape:", test_data.shape)

# Initialize the rolling training sample with the training data
updated_history = train_data.copy()

# -------------------------------------------------------------
# Step 3: Helper Function to Create Daily Sequences
# -------------------------------------------------------------
def create_daily_sequences(df, features_list, target_column):
    """
    Group the DataFrame by day (assume complete day = exactly 24 rows) and return:
      - X_seq: numpy array of shape (n_days, 24, num_features)
      - y_seq: numpy array of shape (n_days, 24, 1)
      - days: list of dates
    """
    X_seq = []
    y_seq = []
    days = []
    for day, group in df.groupby(df.index.normalize()):
        group = group.sort_index()
        if len(group) == 24:
            X_seq.append(group[features_list].values)
            y_seq.append(group[target_column].values.reshape(-1, 1))
            days.append(day)
    return np.array(X_seq), np.array(y_seq), days

# -------------------------------------------------------------
# Step 4: Define LSTM Model Builder Function
# -------------------------------------------------------------
def build_lstm_model(input_shape):
    """
    Build an LSTM model that takes input of shape (lookback, num_features)
    and outputs a sequence of shape (lookback, 1) i.e. one prediction per time step.
    """
    model = Sequential()
    model.add(LSTM(80, return_sequences=True, input_shape=input_shape))
    model.add(Dropout(0.2))
    model.add(LSTM(40, return_sequences=True))
    model.add(Dropout(0.2))
    model.add(LSTM(30, return_sequences=True))
    # Use TimeDistributed(Dense(1)) so that each time step yields a single prediction.
    model.add(TimeDistributed(Dense(1)))
    model.compile(optimizer="adam", loss="mse")
    return model

# Set lookback and horizon (for daily sequences, lookback = 24 hours, horizon = 24)
lookback = 24
horizon = 24  # we produce 24 predictions per day
num_features = len(features_list)
input_shape = (lookback, num_features)

# -------------------------------------------------------------
# Step 5: Rolling Forecast Loop (Re-train LSTM Each Day)
# -------------------------------------------------------------
# For each test day (24 hourly observations), we will:
#   a) Re-fit scalers on updated_history.
#   b) Create daily training sequences from updated_history.
#   c) Re-train an LSTM model on these scaled sequences.
#   d) Scale the test day's raw features and reshape to (1, 24, num_features).
#   e) Forecast a sequence of 24 hourly predictions, inverse-transform them.
#   f) Append the test day's raw features with the forecasted target to updated_history (rolling the window).
forecast_results = pd.DataFrame()

# Get unique test days (using date portion only)
test_days = pd.to_datetime(test_data.index.date).unique()
test_days = np.sort(test_days)
print("Number of forecast days in test period:", len(test_days))

# Increase the number of epochs per day; you can adjust this parameter.
epochs_per_step = 50

for day in test_days:
    # (a) Extract raw features for the test day (should be 24 rows)
    day_mask = test_data.index.normalize() == pd.Timestamp(day)
    X_test_day_raw = test_data.loc[day_mask, features_list].copy()

    # Skip if incomplete day
    if X_test_day_raw.empty or len(X_test_day_raw) < 24:
        print(f"Insufficient data for {day}. Skipping.")
        continue
    X_test_day_raw = X_test_day_raw.sort_index()

    # (b) Re-fit scalers on the current updated_history
    scaler_X = MinMaxScaler(feature_range=(-1,1))
    scaler_y = MinMaxScaler(feature_range=(-1,1))
    scaler_X.fit(updated_history[features_list])
    scaler_y.fit(updated_history[[target_column]])

    # (c) Create daily training sequences from updated_history (only use complete days)
    X_train_seq, y_train_seq, train_days_list = create_daily_sequences(updated_history, features_list, target_column)
    if len(X_train_seq) == 0:
        print(f"No complete training sequences available. Skipping forecast for {day}.")
        continue
    # Scale each daily sequence using the current scalers
    X_train_scaled = np.array([scaler_X.transform(x) for x in X_train_seq])  # shape (n_days, 24, num_features)
    y_train_scaled = np.array([scaler_y.transform(y) for y in y_train_seq])      # shape (n_days, 24, 1)

    # (d) Build the LSTM model using input shape (24, num_features)
    lstm_model = build_lstm_model(input_shape)

    # Train the LSTM model on the current training sequences
    lstm_model.fit(X_train_scaled, y_train_scaled, epochs=epochs_per_step, batch_size=32, verbose=0)

    # (e) Prepare the test day's input: scale and reshape to (1, 24, num_features)
    X_test_day_scaled = scaler_X.transform(X_test_day_raw)
    X_test_day_seq = np.expand_dims(X_test_day_scaled, axis=0)

    # Predict: result shape is (1, 24, 1)
    y_forecast_scaled = lstm_model.predict(X_test_day_seq)
    # Inverse-transform predictions; result shape will be (1, 24) after flattening the last dimension.
    y_forecast = scaler_y.inverse_transform(y_forecast_scaled.reshape(-1, 1)).reshape(1, -1)

    # (f) Create a DataFrame for the forecasted day (24 hourly predictions)
    df_forecast = pd.DataFrame({target_column: y_forecast.flatten()}, index=X_test_day_raw.index)

    # (g) Update the training sample (rolling window) with the forecasted day;
    # combine the raw features for the test day with the forecasted target.
    new_training_rows = X_test_day_raw.copy()
    new_training_rows[target_column] = df_forecast[target_column]
    updated_history = pd.concat([updated_history, new_training_rows]).sort_index()

    # Append the day's forecast to the overall forecast results
    forecast_results = pd.concat([forecast_results, df_forecast])

    print(f"Forecast completed for {day}.")

print("Rolling forecast completed.")
print("Forecast results shape:", forecast_results.shape)

# -------------------------------------------------------------
# Step 6: Evaluation and Visualization
# -------------------------------------------------------------
# Align the forecast results with the actual test targets (if available)
forecast_results = forecast_results.sort_index()
actual_test = test_data[[target_column]].loc[forecast_results.index]

# Combine actual and forecasted values for evaluation
results = actual_test.copy()
results['predicted_price'] = forecast_results[target_column]

# Compute Mean Squared Error (MSE)
mse = mean_squared_error(results[target_column], results['predicted_price'])
print(f"Test MSE: {mse:.2f}")

# Plot Actual vs. Predicted prices
plt.figure(figsize=(28, 9))
plt.xlabel('Date', fontsize=20)
plt.ylabel('Euro/mWh', fontsize=20)
plt.plot(results[target_column], label='Actual', color='r')
plt.plot(results['predicted_price'], label='Predicted', color='g')
plt.tick_params(axis='y', labelsize=20)
plt.legend(prop={'size': 20})
plt.title('Electricity Price Forecast vs Actual (Rolling LSTM)', fontsize=24)
plt.show()

# -------------------------------------------------------------
# Step 7: Save Forecast Results
# -------------------------------------------------------------
results.to_csv('Electricity_Price_Forecast_Results_LSTM_Rolling.csv', header=True,
               quoting=csv.QUOTE_NONE, escapechar=' ')
results.to_excel('Electricity_Price_Forecast_Results_LSTM_Rolling.xlsx', header=True)
print("Forecast results saved to CSV and Excel.")
```

```python
# -------------------------------------------------------------
# Step 0: Imports and Reproducibility Settings
# -------------------------------------------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import datetime
import csv
import warnings

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, TimeDistributed
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error

# Optionally, suppress warnings about feature names if desired:
# warnings.filterwarnings("ignore", category=UserWarning)

# Set seeds for reproducibility
np.random.seed(42)
tf.random.set_seed(42)

# -------------------------------------------------------------
# Step 1: Data Loading and Preprocessing
# -------------------------------------------------------------
# Load your Excel file (adjust the file path as needed)
data = pd.read_excel('/content/EPF_data_Indiv.xlsx')

# Expected columns: Date, Price, Demand Day-ahead DE, Wind and PV Day ahead (MWh/h), Gas, Coal, CO2
print("Raw data:")
print(data.head())

# Remove duplicate Date rows and convert "Date" to datetime
data.drop(data.loc[data['Date'].duplicated()].index, inplace=True)
data['Date'] = pd.to_datetime(data['Date'])

# Set "Date" as the index and drop the original column
data.set_index('Date', inplace=True)

# Filter to keep only records from 2019-01-01 onward
data = data[data.index >= '2019-01-01']
print("Data after filtering (from 2019-01-01):")
print(data.head())

# (Optional) Add extra time features (e.g., Hour) if needed
# data['Hour'] = data.index.hour

# -------------------------------------------------------------
# Step 2: Define Target, Features, and Split Data by Year
# -------------------------------------------------------------
# According to your file:
#   Target: "Price"
#   Features: ["Demand Day-ahead DE", "Wind and PV Day ahead (MWh/h)", "Gas", "Coal", "CO2"]
target_column = "Price"
features_list = ["Demand Day-ahead DE", "Wind and PV Day ahead (MWh/h)", "Gas", "Coal", "CO2"]
print("Target column:", target_column)
print("Feature columns:", features_list)

# Get unique years in the dataset
unique_years = sorted(data.index.year.unique())
print("Unique years in data:", unique_years)

# Define training sample (all data up to year 2022) and test sample (years > 2022)
train_years = [year for year in unique_years if year <= 2022]
test_years  = [year for year in unique_years if year > 2022]
print("Training years:", train_years)
print("Test years:", test_years)

train_data = data[data.index.year.isin(train_years)]
test_data  = data[data.index.year.isin(test_years)]
print("Training data shape:", train_data.shape)
print("Test data shape:", test_data.shape)

# Initialize our rolling training sample with the original training data.
updated_history = train_data.copy()

# -------------------------------------------------------------
# Step 3: Helper Function to Create Sliding Windows
# -------------------------------------------------------------
def create_sliding_windows(df, n_steps_in, n_steps_out, features_list, target_column):
    """
    Create sliding windows from DataFrame df (assumed sorted by date).
    Each window is of length (n_steps_in + n_steps_out). The function returns:
      - X_windows: numpy array of shape (n_samples, n_steps_in, num_features) for inputs.
      - y_windows: numpy array of shape (n_samples, n_steps_out) for target values.
    """
    X_windows = []
    y_windows = []
    for i in range(len(df) - (n_steps_in + n_steps_out) + 1):
        X_seq = df.iloc[i:i+n_steps_in][features_list].values  # input sequence (n_steps_in, num_features)
        y_seq = df.iloc[i+n_steps_in:i+n_steps_in+n_steps_out][target_column].values  # target sequence (n_steps_out,)
        X_windows.append(X_seq)
        y_windows.append(y_seq)
    return np.array(X_windows), np.array(y_windows)

# -------------------------------------------------------------
# Step 4: Define LSTM Model Builder Function
# -------------------------------------------------------------
def build_lstm_model(input_shape, output_horizon):
    """
    Build an LSTM model that takes input of shape (n_steps_in, num_features)
    and outputs a sequence of length output_horizon, one value per time step.
    """
    model = Sequential()
    model.add(LSTM(80, return_sequences=True, input_shape=input_shape))
    model.add(Dropout(0.2))
    model.add(LSTM(40, return_sequences=True))
    model.add(Dropout(0.2))
    model.add(LSTM(30, return_sequences=True))
    # TimeDistributed layer wraps Dense(1) so each output step gets one prediction.
    model.add(TimeDistributed(Dense(1)))
    model.compile(optimizer="adam", loss="mse")
    return model

# Set parameters: Lookback = 168 (one week), Forecast Horizon = 24 (next day)
n_steps_in = 168
n_steps_out = 24
num_features = len(features_list)
input_shape = (n_steps_in, num_features)

# -------------------------------------------------------------
# Step 5: Rolling Forecast Loop (Re-train LSTM Each Day)
# -------------------------------------------------------------
# For each forecast day, we will:
#   (a) Re-fit scalers on updated_history.
#   (b) Create sliding windows (using all available hourly data in updated_history).
#   (c) Train the LSTM model on the latest training windows.
#   (d) Use the last n_steps_in hours from updated_history as input to forecast the next 24 hours.
#   (e) Inverse-transform the forecast to original scale.
#   (f) Update updated_history by appending the test day's raw features with the forecasted target.
forecast_results = pd.DataFrame()

# Get unique test days (based on the date portion)
test_days = pd.to_datetime(test_data.index.date).unique()
test_days = np.sort(test_days)
print("Number of forecast days in test period:", len(test_days))

# Set number of epochs per rolling step (you can adjust as needed)
epochs_per_step = 50

for day in test_days:
    # (a) For the current test day, extract its raw exogenous features.
    # Since our forecast horizon is 24 hours, we expect 24 rows in the test day.
    day_mask = test_data.index.normalize() == pd.Timestamp(day)
    X_test_day_raw = test_data.loc[day_mask, features_list].copy()

    # Skip this day if fewer than 24 rows exist
    if X_test_day_raw.empty or len(X_test_day_raw) < n_steps_out:
        print(f"Insufficient data for {day}. Skipping.")
        continue
    X_test_day_raw = X_test_day_raw.sort_index()

    # (b) Re-fit scalers on updated_history
    scaler_X = MinMaxScaler(feature_range=(-1,1))
    scaler_y = MinMaxScaler(feature_range=(-1,1))
    scaler_X.fit(updated_history[features_list])
    scaler_y.fit(updated_history[[target_column]])

    # (c) Create sliding windows from updated_history.
    X_train_windows, y_train_windows = create_sliding_windows(updated_history, n_steps_in, n_steps_out, features_list, target_column)
    if len(X_train_windows) == 0:
        print(f"No complete training windows available for {day}. Skipping forecast.")
        continue
    # Scale the windows using the fitted scalers.
    X_train_scaled = np.array([scaler_X.transform(x) for x in X_train_windows])         # shape: (n_samples, 168, num_features)
    y_train_scaled = np.array([scaler_y.transform(y.reshape(-1, 1)).flatten() for y in y_train_windows])  # shape: (n_samples, 24)

    # Reshape y_train_scaled to (n_samples, 24, 1) for model training.
    y_train_scaled = y_train_scaled.reshape(y_train_scaled.shape[0], y_train_scaled.shape[1], 1)

    # (d) Build and train the LSTM model using the current training windows.
    lstm_model = build_lstm_model(input_shape, n_steps_out)
    lstm_model.fit(X_train_scaled, y_train_scaled, epochs=epochs_per_step, batch_size=32, verbose=0)

    # (e) Forecast for the test day:
    # Use the most recent n_steps_in hours from updated_history as input.
    X_input = updated_history[features_list].iloc[-n_steps_in:]
    # Ensure X_input is a DataFrame with proper column names.
    X_input = pd.DataFrame(X_input, columns=features_list)
    X_input_scaled = scaler_X.transform(X_input)
    X_input_scaled = X_input_scaled.reshape(1, n_steps_in, num_features)

    # Predict: output shape will be (1, n_steps_out, 1)
    y_forecast_scaled = lstm_model.predict(X_input_scaled)
    y_forecast = scaler_y.inverse_transform(y_forecast_scaled.reshape(-1, 1)).flatten()  # 24 values

    # (f) Construct a DataFrame for the forecasted 24 hours,
    # using the index of the test day (assume the test day provides exactly 24 hours in order)
    df_forecast = pd.DataFrame({target_column: y_forecast}, index=X_test_day_raw.index)

    # (g) Update the rolling training sample: Append the test day's raw features together with the forecasted Price.
    new_training_rows = X_test_day_raw.copy()
    new_training_rows[target_column] = df_forecast[target_column]
    updated_history = pd.concat([updated_history, new_training_rows]).sort_index()

    # (h) Append the forecast for this test day to the overall forecast results.
    forecast_results = pd.concat([forecast_results, df_forecast])

    print(f"Forecast completed for {day}.")

print("Rolling forecast completed.")
print("Forecast results shape:", forecast_results.shape)

# -------------------------------------------------------------
# Step 6: Evaluation and Visualization
# -------------------------------------------------------------
# Align forecast_results with actual test targets (if available)
forecast_results = forecast_results.sort_index()
actual_test = test_data[[target_column]].loc[forecast_results.index]

# Combine actual and forecasted values
results = actual_test.copy()
results['predicted_price'] = forecast_results[target_column]

# Compute Mean Squared Error (MSE)
mse = mean_squared_error(results[target_column], results['predicted_price'])
print(f"Test MSE: {mse:.2f}")

# Plot Actual vs. Predicted Prices
plt.figure(figsize=(28, 9))
plt.xlabel('Date', fontsize=20)
plt.ylabel('Euro/mWh', fontsize=20)
plt.plot(results[target_column], label='Actual', color='r')
plt.plot(results['predicted_price'], label='Predicted', color='g')
plt.tick_params(axis='y', labelsize=20)
plt.legend(prop={'size': 20})
plt.title('Electricity Price Forecast vs Actual (Rolling LSTM)', fontsize=24)
plt.show()

# -------------------------------------------------------------
# Step 7: Save Forecast Results
# -------------------------------------------------------------
results.to_csv('Electricity_Price_Forecast_Results_LSTM_Rolling.csv', header=True,
               quoting=csv.QUOTE_NONE, escapechar=' ')
results.to_excel('Electricity_Price_Forecast_Results_LSTM_Rolling.xlsx', header=True)
print("Forecast results saved to CSV and Excel.")
```
